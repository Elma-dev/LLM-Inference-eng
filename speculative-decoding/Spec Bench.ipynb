# Helpful Function:

```python
def get_ngram_metrics(llm):
    map_spec = {
        "vllm:spec_decode_num_accepted_tokens":"accepted",
        "vllm:spec_decode_num_draft_tokens":"draft",
        "vllm:spec_decode_num_drafts":"steps",
    }
    metrics = {"accepted": 0, "draft": 0, "steps": 0}
    for x in llm.get_metrics():
        if "spec" in x.name:
            key = map_spec.get(x.name)
            if key:
                try:
                    value = x.value if x.value else 0
                except:
                    value = x.value if x.value else 0
                metrics[key] = value
            else:
                continue
    return metrics
```

```python
# test
get_ngram_metrics(llm)
```

```python
import time
import numpy as np

def get_bench_values(llm, prompts, sampling_params=None):
    pre_metrics = get_ngram_metrics(llm)
    start = time.perf_counter()
    outputs = llm.generate(prompts, sampling_params=sampling_params)
    elapsed = time.perf_counter() - start
    post_metrics = get_ngram_metrics(llm)

    # Per-prompt stats
    per_prompt_tokens = []
    ttfts_ms = []
    decode_times_s = []
    decode_tps_list = []
    total_latencies_s = []
    queue_waits_s = []

    for o in outputs:
        # Output token count
        out_tokens = sum(len(c.token_ids) for c in o.outputs)
        per_prompt_tokens.append(out_tokens)

        m = o.metrics
        if m:
            # TTFT — vLLM computes this across clock types internally
            if m.first_token_latency is not None:
                ttfts_ms.append(m.first_token_latency * 1000)

            # Decode time: first_token → last_token (both monotonic)
            if m.first_token_ts and m.last_token_ts:
                decode_time = m.last_token_ts - m.first_token_ts
                decode_times_s.append(decode_time)
                if decode_time > 0:
                    decode_tps_list.append(out_tokens / decode_time)

            # Total request latency (monotonic: queued → last_token)
            if m.queued_ts and m.last_token_ts:
                total_latencies_s.append(m.last_token_ts - m.queued_ts)

            # Queue wait: queued → scheduled (both monotonic)
            if m.queued_ts and m.scheduled_ts:
                queue_waits_s.append(m.scheduled_ts - m.queued_ts)

    total_tokens = sum(per_prompt_tokens)
    wall_tps = total_tokens / elapsed if elapsed > 0 else 0.0

    # Speculation deltas
    delta_accepted = post_metrics["accepted"] - pre_metrics["accepted"]
    delta_draft    = post_metrics["draft"]    - pre_metrics["draft"]
    delta_steps    = post_metrics["steps"]    - pre_metrics["steps"]
    delta_rejected = delta_draft - delta_accepted

    acceptance_rate       = delta_accepted / delta_draft  if delta_draft > 0  else 0.0
    rejection_rate        = delta_rejected / delta_draft  if delta_draft > 0  else 0.0
    avg_accepted_per_step = delta_accepted / delta_steps  if delta_steps > 0  else 0.0

    def stats(lst):
        if not lst:
            return {"mean": None, "p50": None, "p90": None, "p99": None, "min": None, "max": None}
        a = np.array(lst)
        return {
            "mean": round(float(np.mean(a)), 3),
            "p50":  round(float(np.percentile(a, 50)), 3),
            "p90":  round(float(np.percentile(a, 90)), 3),
            "p99":  round(float(np.percentile(a, 99)), 3),
            "min":  round(float(np.min(a)), 3),
            "max":  round(float(np.max(a)), 3),
        }

    return {
        # ── Throughput ────────────────────────────────────────────
        "wall_tps":               round(wall_tps, 2),
        "decode_tps":             stats(decode_tps_list),        # excludes prefill
        "total_tokens":           total_tokens,
        "total_time_s":           round(elapsed, 3),
        "num_prompts":            len(prompts),

        # ── Latency (ms) ─────────────────────────────────────────
        "ttft_ms":                stats([t for t in ttfts_ms]),
        "decode_time_ms":         stats([t * 1000 for t in decode_times_s]),
        "total_latency_ms":       stats([t * 1000 for t in total_latencies_s]),
        "queue_wait_ms":          stats([t * 1000 for t in queue_waits_s]),

        # ── Output token distribution ─────────────────────────────
        "output_tokens":          stats(per_prompt_tokens),

        # ── Speculation ───────────────────────────────────────────
        "speculation_active":     delta_draft > 0,
        "acceptance_rate":        round(acceptance_rate, 4),
        "rejection_rate":         round(rejection_rate, 4),
        "avg_accepted_per_step":  round(avg_accepted_per_step, 3),
        "delta_draft_tokens":     delta_draft,
        "delta_accepted_tokens":  delta_accepted,
        "delta_rejected_tokens":  delta_rejected,
        "delta_steps":            delta_steps,
    }
```

```python
# test
get_bench_values(llm,["hi"],sampling_params)
```

# Analyzing

```python
import pandas as pd

dfs = []
for config in ["base", "ngram","mtp"]:
    for task in ["code", "summarization"]:
        df = pd.read_parquet(f"{config}_{task}.parquet")
        df["config"] = config
        df["task"] = task
        dfs.append(df)

df = pd.concat(dfs, ignore_index=True)
```

```python
df.head()
```

```python
df.shape
```

```python
# Assuming you have a 'config' column (baseline / ngram)
# and a 'task' column from 'turns'

summary = df.groupby(["task", "config"]).agg(
    wall_tps=("wall_tps", "mean"),
    decode_tps=("decode_tps", lambda x: np.mean([v["mean"] for v in x])),
    ttft_p90=("ttft_ms", lambda x: np.mean([v["p90"] for v in x])),
    acceptance_rate=("acceptance_rate", "mean"),
    avg_accepted_per_step=("avg_accepted_per_step", "mean"),
).round(3)

print(summary)
```

```python
baseline = df[df["config"] == "base"].groupby("task")["wall_tps"].mean()
ngram    = df[df["config"] == "ngram"].groupby("task")["wall_tps"].mean()
speedup  = (ngram / baseline).round(3)
print(speedup)
```

```python
baseline = df[df["config"] == "base"].groupby("task")["wall_tps"].mean()
mtp    = df[df["config"] == "mtp"].groupby("task")["wall_tps"].mean()
speedup  = (mtp / baseline).round(3)
print(speedup)
```
